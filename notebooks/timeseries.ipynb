# Redis Timeseries Demo

```redis
%connect -h localhost -p 6379
```

## Remove the keys for the timeseries in case they exist already

```redis
DEL ts:carbon_monoxide
DEL ts:relative_humidity
DEL ts:temperature RETENTION 60 LABELS sensor_id 2 area_id 32
```

## Create the timeseries

```redis
TS.CREATE ts:carbon_monoxide
TS.CREATE ts:relative_humidity
TS.CREATE ts:temperature RETENTION 60 LABELS sensor_id 2 area_id 32
```

## Load the data

```redis
%load data/air-quality.txt --quiet
```

## Query the data

```redis
TS.GET ts:temperature
```

```redis
TS.MGET FILTER area_id=32
```

```redis
TS.MGET WITHLABELS FILTER area_id=32
```
